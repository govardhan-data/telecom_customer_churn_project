```python
import pandas as pd 
import numpy as np
import matplotlib as plt
import seaborn as sns
print("library imported")
```

```python
df = pd.read_csv(r'c:\users\hardy\telecom_customer_churn_project\data\raw\telecomdata.csv')
print("data loaded ")
print("shape:" , df.shape)
df.head()
```

```python
df.info()
print("\n missingvalues:\n",df.isnull().sum())
```

```python
print("Churn distributin (%):")
print(df['Churn'].value_counts(normalize=True)*100)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
sns.countplot(x='Churn' , data=df)
plt.title(" Telecom ustomer churn Distribution")

plt.subplot(1,2,2)
sns.histplot(data = df, x='MonthlyCharges', hue='Churn' , kde = True)
plt.title("MOnthly charges ")
plt.tight_layout()
plt.show()
```

```python
print (" Churn by COntract :")
print (df.groupby('Contract')['Churn'].value_counts(normalize=True)*100)
```
